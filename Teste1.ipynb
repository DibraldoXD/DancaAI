```python
import cv2
import mediapipe as mp
import numpy as np
import pygame
import time
from mediapipe import solutions

mp_drawing = mp.solutions.drawing_utils
mp_pose = mp.solutions.pose
```

```python
# --- CONFIGURAÇÕES DA MÚSICA ---
BPM = 147 # Coloque o BPM exato da sua música aqui
SEGUNDOS_POR_BATIDA = 60.0 / BPM
ARQUIVO_MUSICA = 'musica_teste.mp3'

# Inicializa o mixer do Pygame e carrega a música
pygame.mixer.init()
pygame.mixer.music.load(ARQUIVO_MUSICA)
```

```python
cap = cv2.VideoCapture(5)

# Define a largura (Property ID 3)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1920) 

# Define a altura (Property ID 4)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 1080)

# Curl counter variables
counter = 0 
stage = None

## Setup mediapipe instance
with mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5) as pose:

    # Dá o play na música e marca o tempo inicial exato
    pygame.mixer.music.play()
    tempo_inicio_musica = time.time()

    while cap.isOpened():
        ret, frame = cap.read()

        # Calcula o tempo atual da música
        tempo_atual = time.time() - tempo_inicio_musica
        # Descobre em qual batida geral a música está (ex: batida 15)
        batida_atual_total = int(tempo_atual / SEGUNDOS_POR_BATIDA)
        # Converte para o compasso de 4 tempos (1, 2, 3, 4)
        tempo_do_compasso = (batida_atual_total % 4) + 1
        
        # Recolor image to RGB
        image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image.flags.writeable = False
      
        # Make detection
        results = pose.process(image)
    
        # Recolor back to BGR
        image.flags.writeable = True
        image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
        
        # Extract landmarks
        try:
            landmarks = results.pose_landmarks.landmark
            
            # Get coordinates
            kneey1 = landmarks[mp_pose.PoseLandmark.LEFT_KNEE.value].y
            kneey2 = landmarks[mp_pose.PoseLandmark.RIGHT_KNEE.value].y
            
            limiar_transferencia = 0.0007
            mudou_de_peso = False
            
            # Curl counter logic
            if counter == 4:
                cv2.putText(image, '3 REPS COMPLETED', (15,60), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2, cv2.LINE_AA)
                counter = 1
            
            # Curl counter logic
            if kneey1 > kneey2 + limiar_transferencia:
                if stage != "esquerda":
                   counter +=1     
                stage = "esquerda"
                mudou_de_peso = True
                print(counter)

            if kneey1 + limiar_transferencia < kneey2 :
                if stage != "direita":
                   counter +=1    
                stage = "direita"
                mudou_de_peso = True
                print(counter)
                       
        except:
            pass
        
        # Setup status box
        cv2.rectangle(image, (0,0), (225,73), (245,117,16), -1)
        
        # --- RENDERIZAÇÃO NA TELA ---
        cv2.rectangle(image, (0,0), (350,100), (245,117,16), -1)
        
        # Mostra em qual tempo da música estamos rolando (1, 2, 3 ou 4)
        cv2.putText(image, f'TEMPO DA MUSICA: {tempo_do_compasso}', (10,30), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255,255,255), 2, cv2.LINE_AA)
        
        cv2.putText(image, f'PASSOS: {counter}', (10,70), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255,255,255), 2, cv2.LINE_AA)
        
        # Render detections
        mp_drawing.draw_landmarks(image, results.pose_landmarks, mp_pose.POSE_CONNECTIONS,
                                    mp_drawing.DrawingSpec(color=(245,117,66), thickness=2, circle_radius=2), 
                                    mp_drawing.DrawingSpec(color=(245,66,230), thickness=2, circle_radius=2)
        )
        
        cv2.imshow('Mediapipe Feed - Avaliador de Ritmo', image)

        if cv2.waitKey(10) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
    pygame.mixer.music.stop()
```
