```python
import cv2
import mediapipe as mp
import numpy as np
from mediapipe import solutions

mp_drawing = mp.solutions.drawing_utils
mp_pose = mp.solutions.pose
```

```python
cap = cv2.VideoCapture(0)

# Define a largura (Property ID 3)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1920) 

# Define a altura (Property ID 4)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 1080)

# Curl counter variables
counter = 0 
stage = None

## Setup mediapipe instance
with mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5) as pose:
    while cap.isOpened():
        ret, frame = cap.read()
        
        # Recolor image to RGB
        image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        image.flags.writeable = False
      
        # Make detection
        results = pose.process(image)
    
        # Recolor back to BGR
        image.flags.writeable = True
        image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)
        
        # Extract landmarks
        try:
            landmarks = results.pose_landmarks.landmark
            
            # Get coordinates
            kneey1 = landmarks[mp_pose.PoseLandmark.LEFT_KNEE.value].y
            kneey2 = landmarks[mp_pose.PoseLandmark.RIGHT_KNEE.value].y
            footx1 = landmarks[mp_pose.PoseLandmark.LEFT_FOOT_INDEX.value].x
            footx2 = landmarks[mp_pose.PoseLandmark.RIGHT_FOOT_INDEX.value].x
            footz1 = landmarks[mp_pose.PoseLandmark.LEFT_FOOT_INDEX.value].z
            footz2 = landmarks[mp_pose.PoseLandmark.RIGHT_FOOT_INDEX.value].z
            
            if counter == 4:
                cv2.putText(image, '3 REPS COMPLETED', (15,60), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2, cv2.LINE_AA)
                counter = 1
            
            # Curl counter logic
            if kneey1 > kneey2 + 0.0007:
                if stage != "esquerda":
                   counter +=1     
                stage = "esquerda"
                print(counter)
            if kneey1 + 0.0007 < kneey2 :
                if stage != "direita":
                   counter +=1    
                stage = "direita"
                print(counter)
                       
        except:
            pass
        
        # Render curl counter
        # Setup status box
        cv2.rectangle(image, (0,0), (225,73), (245,117,16), -1)
        
        # Rep data
        cv2.putText(image, 'REPS', (15,12), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,0,0), 1, cv2.LINE_AA)
        cv2.putText(image, str(counter), 
                    (10,60), 
                    cv2.FONT_HERSHEY_SIMPLEX, 2, (255,255,255), 2, cv2.LINE_AA)
        
        # Stage data
        cv2.putText(image, 'STAGE', (65,12), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,0,0), 1, cv2.LINE_AA)
        cv2.putText(image, stage, 
                    (60,60), 
                    cv2.FONT_HERSHEY_SIMPLEX, 2, (255,255,255), 2, cv2.LINE_AA)
        
        
        # Render detections
        mp_drawing.draw_landmarks(image, results.pose_landmarks, mp_pose.POSE_CONNECTIONS,
                                mp_drawing.DrawingSpec(color=(245,117,66), thickness=2, circle_radius=2), 
                                mp_drawing.DrawingSpec(color=(245,66,230), thickness=2, circle_radius=2) 
                                 )               
        
        cv2.imshow('Mediapipe Feed', image)

        if cv2.waitKey(10) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```
